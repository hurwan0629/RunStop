```python
import osmnx as ox

print(ox.__version__)
```

```python
# 다운받을 지도의 중심
center = (37.5665, 126.9780)

# 데이터 다운로드 하기
G = ox.graph.graph_from_point(
  center_point=center,
  dist=1000, # km
  network_type="walk" # 도로망 형식
)

print(G)
```

```python
# G.node 출력해보기
# 내부에 들어있는 정보를 볼 수 있다.
for node_id, data in list(G.nodes(data=True))[:5]:
  print(node_id)
  print(data)
  print()
```

for u, v, key, data in list(G.edges(keys=True, data=True))[:5]:
    print("출발:", u)
    print("도착:", v)
    print(data)
    print()

```python
# edge 데이터 출력해보기
# 데이터도 함께 받았을 때 시작/종료 노드와 그 두 노드를 잇는 key(인덱스)
# 그리고 data값이 나오는 것을 알 수 있다.
# 퀄리티가 꽤 좋다.
for u, v, key, data in list(G.edges(keys=True, data=True))[:5]:
    print("출발:", u)
    print("도착:", v)
    print(data)
    print()
```

```python
# 행렬로 데이터 출력해보기
nodes, edges = ox.convert.graph_to_gdfs(G)

# 노드 데이터들 DF
nodes.head()
```

```python
# 거리 데이터 행렬
edges.head()
```

```python
# 공원 데이터 받아오기
parks = ox.features.features_from_point(
    center_point=center,
    tags={
        "leisure": "park" # 태그를 이용해서 원하는 피처 가져오기
    },
    dist=1000 # 1000m 반경 가져오기
)

print(parks.columns)
parks.head()
```
